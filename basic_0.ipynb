```python
# Handling Missing values
num_cols = df_2.select_dtypes(include = ['int64','float64'])
cat_cols = df_2.select_dtypes(include = ['object'])
for col in num_cols:
  df_2[col].fillna(df_2[col].mean(),inplace = True)
for col in cat_cols:
  df_2[col].fillna(df_2[col].mode()[0],inplace = True)
```

```python
df_2['Date'] = pd.to_datetime(df_2['Date'])
```

```python
print(f"Missing Values : {df_2.isnull().sum()}")

print(f"Dupliacted Value : {df_2.duplicated().sum()}")
```

```python
df_2.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for col in cat_cols:
  df_2[col] = le.fit_transform(df_2[col])
df_2.head()
```

```python
df_2["rain_3day"] = df_2["Rainfall"].rolling(3).sum()
df_2["rain_7day"] = df_2["Rainfall"].rolling(7).sum()
df_2["Soil_wetness_index"] = df_2['Rainfall'] - df_2['Evaporation']
df_2["Pressure_change"] = df_2["Pressure3pm"] - df_2['Pressure9am']
df_2["Humidity_change"] = df_2['Humidity3pm'] - df_2['Humidity9am']
```

```python
# Predict Rain Intensity
X_train = df_2.drop(columns = ["Rainfall","Date","Location","WindDir9am","WindDir3pm","WindGustDir","RainToday","RainTomorrow","Temp3pm","Humidity3pm","Pressure3pm","MaxTemp"])
y_train = df_2["Rainfall"]
from xgboost import XGBRegressor
xgb_model = XGBRegressor()
xgb_model.fit(X_train,y_train)

importances = pd.Series(xgb_model.feature_importances_,index = X_train.columns)
importance = importances.sort_values(ascending = False)
importance
```
